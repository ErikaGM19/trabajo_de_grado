### Ajuste de hiperparámetros - Learning Rate = 0.01

```python
import sys, os
sys.path.append(os.path.abspath('..'))
from experimento_utils import ejecutar_experimento

CARPETA_EXPERIMENTO = os.path.abspath('.')
```

```python
resultado, history = ejecutar_experimento(
    nombre_experimento='resnet50_TL_lr01',
    carpeta_experimento=CARPETA_EXPERIMENTO,
    learning_rate=0.01,
    optimizador='adam',
    batch_size=32,
    dropout=0.5,
    capas_descongeladas=0,
    epochs=30, patience_es=5, patience_lr=3
)
resultado
```

```python
import matplotlib.pyplot as plt
plt.plot(history.history['accuracy'], label='Train')
plt.plot(history.history['val_accuracy'], label='Validation')
plt.title('Accuracy — resnet50_TL_lr01')
plt.xlabel('Epoch'); plt.ylabel('Accuracy'); plt.legend()
plt.show()
```
