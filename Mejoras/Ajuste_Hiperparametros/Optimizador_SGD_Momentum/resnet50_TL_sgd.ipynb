```python
import sys, os
sys.path.append(os.path.abspath('..'))
from experimento_utils import ejecutar_experimento
from mejores_valores import MEJOR_DROPOUT, MEJOR_CAPAS, MEJOR_BATCH

CARPETA_EXPERIMENTO = os.path.abspath('.')
```

```python
resultado, history = ejecutar_experimento(
    nombre_experimento='resnet50_TL_sgd',
    carpeta_experimento=CARPETA_EXPERIMENTO,
    learning_rate=1e-5,
    optimizador='sgd',
    batch_size=MEJOR_BATCH,
    dropout=MEJOR_DROPOUT,
    capas_descongeladas=MEJOR_CAPAS,
    epochs=30, patience_es=5, patience_lr=3
)
resultado
```
